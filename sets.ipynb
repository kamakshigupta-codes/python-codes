```python
#CREATING A SET
my_set = {1,2,3,4,5}
print(my_set)
print(type(my_set))

my_empty_set = set()
print(type(my_empty_set))

my_empty_set = set([1,2,3,4,5,6])
print(my_empty_set)

my_set = set([1,2,3,4,5,6,8])
print(my_set)
```

```python
## BASIC OPERATIONS ON SET 
#ADDING AND REMOVING ELEMENTS 
my_set.add(7)
print(my_set)

#my_set.remove(3)
print(my_set)

my_set.discard(11)
print(my_set)

removed_elements = my_set.pop()
print(removed_elements)
print(my_set)

my_set.clear()
print(my_set)

my_set = {1,2,3,4,5}
print(3 in my_set)
print(10 in my_set)




```

```python
## MATHEMATICAL OPERATION 
set1 = {1,2,3,4,5,6}
set2 = {4,5,6,7,8,9}

#UNION
union_set = set1.union(set2)
print(union_set)

#INTERSECTION 
intersection_set = set1.intersection(set2)
print(intersection_set)

#INTERSECTION UPDATE
set.intersection_update(set2)
print(set1)

#FIND + UPDATE ORIGINAL SET
set1 = {1,2,3,4,5,6}
set2 = {4,5,6,7,8,9}

#DIFFERENCE 
print(set1.difference(set2))

set2.difference(set1)

#SYMMENTRIC DIFFERENCE
set1.symmetric_difference(set2)

##sets method 
set1 = {1,2,3}
set2 = {3,4,5}

##is subset
print(set1.issubset(set2))

##is superset
print(set1.issuperset(set2))
```
