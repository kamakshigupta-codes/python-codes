```python
#Understanding variable types 
age = 25 #integer
height = 6.1 #float
name = "Kamakshi" #str
is_student = True #boolean 

print(type(age))
print(type(height))
print(type(name))
type(name)
type(height)
type(is_student)
```

```python
#Type conversion 
age_str = str(age)
print(age_str)
print(type(age_str))
```

```python
#Example simple calculator
num1 = float(input("Enter first number:"))
num2 = float(input("Enter second number:"))
sum = num1+num2
difference = num1-num2
product = num1*num2
quotient = num1/num2
print("sum:",sum)
print("difference:",difference)
print("product:",product)
print("quotient:",quotient)
```
