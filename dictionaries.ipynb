```python
## Creating dictionaries 
empty_dict = {}
print(type(empty_dict))

empty_dict = dict()
empty_dict

student = {"name": "Krish","age":32,"grade":24}
print(student)

student = {"name": "Krish","age":32,"name":24}
print(student)

```

```python
#accessing dictionary elements 
student = {"name":"Krish","age":32,"grade":"A"}
print(student)
print(student["grade"])
print(student["age"])

print(student["grade"])
print(student["last_name"])
print(student["last_name","not available"])
```

```python
## Modify dictionary elements 
student ["age"] = 33 #update value for the key
print(student)
```

```python
del student ['grade']
print(student)
```

```python
## Dictionary methods 
keys = student.keys() #get all the keys
print(keys)
values = student.values() #get all values
print(values)

items = student.items() #get all key values pairs
print(items)
```

```python
#Shallow copy 
student_copy = student 
print(student)
print(student_copy)

student['name'] = "krish2"
print(student)
print(student_copy)
```

```python
#Iterating over dictionaries 
#You can use loops to iterating over dictionary , keys, values or items 
#Iterating over keys
for keys in student.keys():
    print(keys)


#Itersting over valus 
for values in student.values():
    print(values)

## Iterate over key value pair
for key,value in student.items():
    print(f"{keys}:{values}")
```

```python
##Nested dictionaries 
students = {
    "student1" : {'name':'krish','age':32 } ,
    "student2" :{'name':"peter","age":35}
}
print(students)
```

```python
#access nested dictionaries elements 
print(students["student2"]["name"])
print(students["student2"]['age'])

```

```python
#Iterating over nested dictionaries 
for student_id,student_info in student.item():
    print(f"{student_id}:{student_info}")
    for key,value in  student_info.items():
        print(f"{key}:{value}")
```

```python
## Dictionary comphrehension
squares = {x:x**2 for x in range(5)}
print(squares)
```

```python
#condition dictionary comprehension 
even = {x:x**2 for x in range (10)
        if x%2 == 0}
print(even) 
```

```python
# Practical example 
number = [1,2,2,3,3,3,4,4,4,4]
frequency = {}

for number in number:
    if number in frequency:
        frequency[number] +=1
    else:
        frequency[number] = 1

print(frequency)    
```

```python
#Merge 2 dictionaries into one 
dict1 = {"a":1 ,"b":2}
dict2 = {"b":3,"c":4}
merge_dict = {**dict1,**dict2}
print(merge_dict)
```
