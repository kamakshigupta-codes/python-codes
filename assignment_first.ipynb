```python
""" Write a program to reverse a three-digit number.

Example:

Input: 123
Output: 321"""

num = int(input("Enter a three digit number:"))
hundreds = num//100
tens = (num //100)%10
ones = num%10

reverse = ones*100 + tens*10 +hundreds
print("reversed number:",reverse)

```

```python
""" Write a program that takes a four-digit number and calculates the sum of its digits.

Example:

Input: 1234
Output: 10"""

num = int(input("Enter a four- digit number:"))
d1 = num//1000
d2 = (num//100)%10
d3 = (num//10)%10
d4 = num%10

sum_digits = d1+d2+d3+d4
print("Sum of digits:",sum_digits)
```
